```python
# Importa las fuentes del proyecto antes de cargar la biblioteca de visualizaciones.
from pathlib import Path
from matplotlib import font_manager

raiz = Path.cwd()
while raiz != raiz.parent and not (raiz / "recursos" / "fuentes").is_dir():
    raiz = raiz.parent
carpeta_fuentes = raiz / "recursos" / "fuentes"
for archivo in ("Lato-Regular.ttf", "Lato-Bold.ttf"):
    ruta_fuente = carpeta_fuentes / archivo
    if not ruta_fuente.is_file():
        raise FileNotFoundError(f"No se encontró la fuente requerida: {ruta_fuente}")
    font_manager.fontManager.addfont(str(ruta_fuente))

# Confirma que la fuente quedó disponible antes de importar y usar la biblioteca.
font_manager.findfont("Lato", fallback_to_default=False)

import pandas as pd
import visualizaciones_inecl as ine

datos = pd.DataFrame({"anio": [2023, 2024, 2025], "valor": [1200, 1350, 1420]})
meta = ine.Metadatos("Cantidad ficticia por año", "Datos ficticios", ficticios=True)
v = ine.grafico_lineas(datos, "anio", "valor", meta)
v.figura
```

```python
# Exporta HTML, CSV y PNG con el motor Matplotlib, sin navegador.
v.guardar_html("salidas/lineas.html")
v.guardar_csv("salidas/lineas.csv")
v.exportar("salidas/lineas.png")
```
